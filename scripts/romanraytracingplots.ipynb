```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from psfsim.romantrace import _RomanRayBundle
```

```python
raybundle_sidelength = 4

rrb = _RomanRayBundle(0.3,0.1,raybundle_sidelength,"J",save_trace_history=True)
```

```python
print(type(rrb.trace_history))
print(len(rrb.trace_history),"",len(rrb.trace_history[0]),"",len(rrb.trace_history[0][0]),"",len(rrb.trace_history[0][0][0]))
```

```python
history = np.array(rrb.trace_history)
print(history.shape)

number_of_intersections=history.shape[0]
```

```python
print(rrb.open)
print("")
print(f"Ray fraction arrived = {np.sum(rrb.open)/(len(rrb.open)**2)}")
```

```python
rays=history.reshape(number_of_intersections,-1,3)
```

```python
print(rays.shape)

number_of_rays = rays.shape[1]
```

```python
plt.figure(figsize=(10,8))

for ray in range(number_of_rays):
    x = rays[:,ray,0]
    z = rays[:,ray,2]
    plt.plot(z,x,marker='o',markersize=5)

plt.xlabel("z (mm)")
plt.ylabel("x (mm)")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(10,8))

for ray in range(number_of_rays):
    y = rays[:,ray,1]
    z = rays[:,ray,2]
    plt.plot(z,y,marker='o',markersize=5)

plt.xlabel("z (mm)")
plt.ylabel("y (mm)")
plt.grid()
plt.show()
```

```python
fig = plt.figure(figsize=(10,8))
ax = fig.add_subplot(111,projection='3d')

for ray in range(number_of_rays):
    x = rays[:,ray,0]
    y = rays[:,ray,1]
    z = rays[:,ray,2]
    ax.plot(x,y,z,marker='o',markersize=5)

ax.set_xlabel("x (mm)")
ax.set_ylabel("y (mm)")
ax.set_zlabel("z (mm)")
plt.grid()
plt.show()
```

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

ray_color = 'navy'

alphas = np.linspace(0.15,1,number_of_intersections)

for ray in range(number_of_rays):
    for ray_segment in range(number_of_intersections):
        x = rays[ray_segment:ray_segment+2,ray,0]
        y = rays[ray_segment:ray_segment+2:,ray,1]
        z = rays[ray_segment:ray_segment+2:,ray,2]
        ax.plot(x,y,z,color=ray_color,alpha=alphas[ray_segment],marker='o',markersize=2)

ax.set_xlabel("x (mm)")
ax.set_ylabel("y (mm)")
ax.set_zlabel("z (mm)")
plt.grid()
plt.show()
```
